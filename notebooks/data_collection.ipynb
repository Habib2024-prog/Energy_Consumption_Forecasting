```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from ucimlrepo import fetch_ucirepo
import os
import warnings
warnings.filterwarnings('ignore')

print("All libraries loaded successfully!")
print(f"Pandas version: {pd.__version__}")
print(f"NumPy version:  {np.__version__}")
```

```python
folders = [
    'data/raw',
    'data/processed',
    'notebooks',
    'src',
    'models',
    'outputs/figures'
]

for folder in folders:
    os.makedirs(f'../{folder}', exist_ok=True)

print("Project folders created:")
for f in folders:
    print(f"  ../{f}/")
```

```python
print("Downloading datasets from UCI Machine Learning Repository...")

datasets= fetch_ucirepo(id=374)
```

```python
X= datasets.data.features
y= datasets.data.targets

df=pd.concat([X,y], axis=1)
print("Dataset loaded successfully!")
print(f"Dataset shape: {df.shape[0]} rows, {df.shape[1]} columns")

```

```python
print("*"*50)
print("Dataset Overview:")
print("*"*50)

print(f"Number of samples: {df.shape[0]}")
print(f"Number of features: {df.shape[1] - 1}")  
print(f"Column names: {df.columns.tolist()}")
print(f"Data types:\n{df.dtypes}")
print(f"First 5 rows:\n{df.head()}")
```

```python
print("="*50)
print("DATA QUALITY CHECK")
print("="*50)

print(f"\nMissing values per column:")
print(df.isnull().sum())

print(f"\nDuplicate rows: {df.duplicated().sum()}")

print(f"\nBasic statistics of target variable (Appliances):")
print(df['Appliances'].describe())
```

```python
# Save raw dataset to file
raw_path = '../data/raw/energy_raw.csv'
df.to_csv(raw_path, index=False)

print(f"Raw dataset saved to: {raw_path}")
print(f"File size: {os.path.getsize(raw_path) / 1024:.1f} KB")
print(f"\nRows: {len(df)}")
print(f"Columns: {len(df.columns)}")
```
